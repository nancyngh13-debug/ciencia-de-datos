Manejo de Datos Faltantes en Pandas: Identificación y Tratamiento

```python
missing_data = retail_data.isna()
print(missing_data)
```

```python
print(missing_data)
```

```python
missing_data_count = retail_data.isna().sum()
print('conteo de datos faltantes por columna:\n', missing_data_count)
```

```python
no_missing_row = retail_data.dropna()
print('Datos sin filas con valores faltantes:\n', no_missing_row)
```

```python
retail_data_fillef_zeros = retail_data.fillna(0)
retail_data_fillef_zeros_count = retail_data.isna().sum()
print(retail_data_fillef_zeros_count)
```

```python
mean_unit_price = retail_data['UnitPrice'].mean()
retail_data['UnitPrice'].fillna(mean_unit_price, inplace=True)
```

```python
mean_unit_price = retail_data['UnitPrice'].mean()
retail_data_filled_mean = retail_data.fillna(mean_unit_price)
retail_data_filled_mean_count = retail_data_filled_mean.isna().sum()
print(retail_data_filled_mean_count)
```
