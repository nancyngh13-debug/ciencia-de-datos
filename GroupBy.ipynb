Group by


```python
import pandas as pd

path = '/content/Online_Retail.csv'
retail_data =  pd.read_csv(path, encoding='latin1')
print(type(retail_data))
print(retail_data)
```

```python
contry_count = retail_data['Country'].value_counts()
print(contry_count)
```

```python
country_group = retail_data.groupby('Country')['Quantity'].sum()
print(country_group)
```

```python
product_sales = retail_data.groupby('Description')['Quantity'].sum().sort_values(ascending=False)
display(product_sales.head(10))
```

```python
display(retail_data.describe())
```

```python
country_stats = retail_data.groupby('Country')['UnitPrice'].agg(['mean', 'min', 'max'])
display(country_stats)
```

```python
country_stock_group = retail_data.groupby(['Country', 'StockCode'])['Quantity'].sum()
print(country_stock_group)
```

```python
def total_revenue(group):
  return group['Quantity'] * group['UnitPrice'].sum()
revenue_per_country = retail_data.groupby('Country').apply(total_revenue)
print(revenue_per_country)
```
