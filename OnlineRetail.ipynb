```python
import pandas as pd
#Leer el dataset en ventas desde un archivo csv
file_path = "/content/Online_Retail.csv"
sales_data = pd.read_csv(file_path, encoding='latin1')

#Convertir la columna IncoiceDate a datetime
sales_data['InvoiceDate'] = pd.to_datetime(sales_data['InvoiceDate'], format='%m/%d/%y %H:%M')

#Eliminar filas con valores faltantes en las columnas criticas
sales_data.dropna(subset=['CustomerID', 'InvoiceDate'], inplace=True)

#Crear una nueva columna 'TotalPrice'
sales_data['TotalPrice'] = sales_data['Quantity'] * sales_data['UnitPrice']

#Mostrar las primeras filas del dataset
print(sales_data.head())
```

```python
#Filtrar ventas en el rein unido
uf_sales = sales_data[sales_data['Country'] == 'United Kingdom']

#Mostrar las primeras filas del dataset de ventas en el Reino Unido
print(uf_sales.head())
```

```python
high_quantity_sales = sales_data[sales_data['Quantity'] > 40]
print(high_quantity_sales.head())
```

```python
uk_high_quantity_sales = sales_data[(sales_data['Country'] == 'United Kingdom') & (sales_data['Quantity'] > 400)]
print(uk_high_quantity_sales.head())
```

```python
sales_2011 = sales_data[sales_data['InvoiceDate'].dt.year == 2011]
print(sales_2011.head())
```

```python
sales_2010 = sales_data[sales_data['InvoiceDate'].dt.year == 2010]
print(sales_2010)
```

```python
sales_dec_2010 = sales_data[(sales_data['InvoiceDate'].dt.year == 2010) & (sales_data['InvoiceDate'].dt.month == 12)]
print(sales_dec_2010)
```
