broadcasting

n NumPy, el broadcasting es una técnica que permite realizar operaciones aritméticas entre arrays de diferentes formas (shapes) sin necesidad de copiarlos o usar bucles explícitos.

Básicamente, NumPy "extiende" (o difunde) automáticamente el array más pequeño para que tenga una forma compatible con el más grande, siguiendo ciertas reglas, y luego aplica la operación elemento a elemento.

🔹 Reglas de Broadcasting
Cuando NumPy compara las dimensiones de dos arrays:

Se alinean las dimensiones desde el final (derecha a izquierda).
En cada dimensión:
Si son iguales, son compatibles.
Si una de ellas es 1, se "estira" para coincidir con la otra.
Si son diferentes y ninguna es 1, no son compatibles y se produce un error.
🔹 Ejemplo básico
Python

Copiar código
import numpy as np

# Array 1D
a = np.array([1, 2, 3])

# Escalar
b = 2

# Broadcasting: b se "expande" para coincidir con a
result = a * b
print(result)  # [2 4 6]

```python
array = np.array([1,2,3,4,5])
print(np.all(array > 0))
print(np.any(array > 4))
```

concatenano array a con array b


```python
array_a = np.array([1,2,3])
array_b = np.array([4,5,6])
concatenated_a = np.concatenate([array_a, array_a])
concatenated_ab = np.concatenate([array_a, array_b])
print(concatenated_a)
print(concatenated_ab)
```
